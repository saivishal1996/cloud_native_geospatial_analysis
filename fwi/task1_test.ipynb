load_and_reproject_polygon

```python
geojson_file_path = 'data/snr1.geojson'
wgs_epsg=4326
utm_epsg=32644
```

```python

gdf = gpd.read_file(geojson_file_path)

# Ensure CRS is set (assume WGS84 if not specified)
if gdf.crs is None:
    gdf.crs = "EPSG:4326"

# Reproject to UTM
gdf_wgs = gdf.to_crs(f"EPSG:{wgs_epsg}")
gdf_utm = gdf.to_crs(f"EPSG:{utm_epsg}")

# Get bounds for STAC query (must remain in WGS84)
bounds_wgs84 = gdf_wgs.bounds.iloc[0]

bbox = (bounds_wgs84.minx, bounds_wgs84.miny, bounds_wgs84.maxx, bounds_wgs84.maxy)

bounds_utm = gdf_utm.bounds.iloc[0]
bbox_utm = (bounds_utm.minx, bounds_utm.miny, bounds_utm.maxx, bounds_utm.maxy)
```

```python
gdf_utm, bbox, bbox_utm
```

```python
start_date = '2024-05-01'
end_date='2025-05-31'
cloud_cover_threshold=30
```

query_sentinel2_data

```python

# Connect to Element84 Earth Search API
catalog = pystac_client.Client.open("https://earth-search.aws.element84.com/v1")
# Search for Sentinel-2 L2A data
search = catalog.search(
collections=["sentinel-2-l2a"],
bbox=bbox,
datetime=f"{start_date}/{end_date}",
query={"eo:cloud_cover": {"lt": cloud_cover_threshold}}
)

items = search.item_collection()
print(f"Found {len(items)} Sentinel-2 scenes")
    
```

```python
items
```

```python
from shapely.geometry import box
from shapely.ops    import transform
import pyproj

# build a transformer to UTM
to_utm = pyproj.Transformer.from_crs(4326, utm_epsg, always_xy=True).transform

lonlat_box = box(*items[0].bbox)
utm_box    = transform(to_utm, lonlat_box)
print(items[0].id, "→ UTM bbox:", utm_box.bounds)
```

```python
data = odc.stac.load(
                [items[0]],
                bands=["red","rededge1", "scl","green","swir16"],
                crs=utm_epsg,
                bbox =bbox,
                resolution=20,
                chunks={'x':2048,'y':2048},
            )
```
